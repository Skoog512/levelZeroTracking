### Monthly Level Zero Report

```python
import pandas as pd
from datetime import datetime
import calendar
```

```python
level_zero_df = pd.read_csv('level_zero.csv')
```

```python
target_month = datetime.now().month - 1
```

```python
level_zero_df['Date'] = pd.to_datetime(level_zero_df['Date'])
level_zero_df_target_month = level_zero_df[pd.to_datetime(level_zero_df['Date']).dt.month == target_month]
```

```python
level_zero_occurrence = len(level_zero_df_target_month) // 2
```

```python
level_zero_df_target_month_group = level_zero_df_target_month.groupby(level_zero_df_target_month.index // 2)["Date"].apply(lambda x: x.iloc[1] - x.iloc[0]).sum()
```

```python
level_zero_df_target_month_group
hours = level_zero_df_target_month_group.seconds // 3600
total_minutes, secs = divmod(level_zero_df_target_month_group.seconds, 60)
total_hours, mins = divmod(total_minutes, 60)
```

```python
level_zero_string = f"Time Spent in Level Zero: {hours} hours {mins} minutes {secs} seconds"
```

```python
num_days = calendar.monthrange(datetime.now().year,target_month)[1]
total_seconds_in_month = num_days * 24 * 60 * 60
total_seconds_in_month
```

```python
percentage_of_time_spent_in_level_zero = round(level_zero_df_target_month_group.seconds / total_seconds_in_month * 100, 3)
```

```python
percentage_of_time_spent_in_level_zero
```

```python
print(f"Level Zero Report {calendar.month_name[target_month]}-{datetime.now().year}\nLevel Zero Occurrence: {level_zero_occurrence}\n{level_zero_string}\nPercent of Time in Level Zero: {percentage_of_time_spent_in_level_zero}% ")
```
